```python
import pandas as pd

# Step 1: Data Loading (Aapke exact paths)
path_forest = r'C:\Users\hp\Desktop\Intership_projects\Bird Species Observation Analysis\Bird_Monitoring_Data_FOREST.XLSX'
path_grassland = r'C:\Users\hp\Desktop\Intership_projects\Bird Species Observation Analysis\Bird_Monitoring_Data_GRASSLAND.XLSX'

# Excel files ko read karna 
excel_forest = pd.ExcelFile(path_forest)
excel_grassland = pd.ExcelFile(path_grassland)

df_list = []

# Forest file ki saari sheets load karna
for sheet in excel_forest.sheet_names:
    sheet_df = pd.read_excel(excel_forest, sheet_name=sheet)
    df_list.append(sheet_df)

# Grassland file ki saari sheets load karna
for sheet in excel_grassland.sheet_names:
    sheet_df = pd.read_excel(excel_grassland, sheet_name=sheet)
    df_list.append(sheet_df)

# Dono files ke data ko ek Master DataFrame mein combine karna
df = pd.concat(df_list, ignore_index=True)
print("Data Successfully Combined! Shape:", df.shape)
```

```python
#Step 2: Data Cleaning & Preprocessing

# Missing categorical values fill karna
if 'Sex' in df.columns:
    df['Sex'] = df['Sex'].fillna('Undetermined')

if 'Disturbance' in df.columns:
    df['Disturbance'] = df['Disturbance'].fillna('No Effect')

# Date columns fix karna
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
    df['Year'] = df['Date'].dt.year
    df['Month'] = df['Date'].dt.month_name()

# Numerical weather metrics fix karna
weather_cols = ['Temperature', 'Humidity']
for col in weather_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].fillna(df[col].mean())

# Cleaned dataset ko save karna
df.to_csv('cleaned_bird_data.csv', index=False)
print("Data Cleaning Complete! Cleaned CSV File saved as 'cleaned_bird_data.csv'.")
```

```python
import pandas as pd
import sqlite3

# Cleaned data load karein
df = pd.read_csv('cleaned_bird_data.csv')

# SQLite database connect/create karein
conn = sqlite3.connect('bird_biodiversity.db')

# Dataframe ko SQL table mein save karein
df.to_sql('bird_observations', conn, if_exists='replace', index=False)

print("Data successfully loaded into SQL Database 'bird_biodiversity.db'!")
```
